```python
def calculate_centipawn_loss(cp_values: list[int], is_white: bool) -> float:
    """
    Calculate average centipawn loss from a list of position evaluations
    
    Args:
        cp_values: List of centipawn evaluations for consecutive positions
        is_white: True if calculating for white, False for black
    
    Returns:
        Average centipawn loss across all moves
    """
    # Group evaluations into pairs of consecutive positions
    pairs = zip(cp_values[:-1], cp_values[1:])
    
    # Calculate loss for each move
    losses = []
    color_multiplier = 1 if is_white else -1
    
    for prev_eval, curr_eval in pairs:
        # Calculate difference between positions
        diff = (curr_eval - prev_eval) * color_multiplier
        # Can't have negative loss (can't gain points for playing better than best move)
        loss = max(0, diff)
        losses.append(loss)
    
    # Return average loss
    print(losses)
    return sum(losses) / len(losses) if losses else 0

# Example usage:
cp_values = [0, 10, 20, 5, 15]   # Example centipawn evaluations
acpl_white = calculate_centipawn_loss(cp_values, is_white=True)
acpl_black = calculate_centipawn_loss(cp_values, is_white=False)

print(f"White's ACPL: {acpl_white}")
print(f"Black's ACPL: {acpl_black}")
```
